# 🔎 Levantamento e Exploração das Fontes de Dados

Tech Challenge — Avaliação da Alfabetização

Este notebook faz parte da etapa de levantamento e exploração das fontes de dados utilizadas no desenvolvimento do projeto.

O objetivo desta etapa é conhecer a estrutura das bases disponíveis antes da implementação dos tratamentos e cruzamentos nas camadas da arquitetura de dados.

## 🎯 Objetivos deste levantamento

Nesta etapa buscamos:

- identificar as fontes de dados disponíveis para o projeto;
- identificar as tabelas, arquivos e estruturas existentes em cada fonte;
- levantar as colunas disponíveis em cada tabela;
- identificar os tipos de dados das variáveis;
- identificar possíveis chaves de relacionamento;
- verificar a cobertura temporal e geográfica das informações;
- identificar quais informações podem contribuir para responder às perguntas de negócio do projeto;
- avaliar a necessidade de fontes complementares.

## 🎯 Pergunta central do projeto

O levantamento das fontes será orientado pela necessidade de compreender o alcance das metas de alfabetização e investigar como os resultados podem se relacionar com características educacionais e socioeconômicas dos municípios e estados.

As perguntas de análise serão refinadas pela equipe à medida que as variáveis disponíveis nas diferentes fontes forem identificadas.

## 🏗️ Relação com a arquitetura do projeto

O levantamento realizado neste notebook antecede a implementação das camadas de dados na AWS.

O fluxo considerado para o projeto é:

**Fontes de dados → Bronze → Prata → Gold → Visualização**

Neste momento, o notebook **não tem como objetivo realizar a modelagem definitiva das camadas**.

A finalidade é entender o conteúdo e a estrutura das fontes para apoiar as decisões que serão tomadas posteriormente pela equipe.

### 🥉 Camada Bronze

A camada Bronze será responsável pelo armazenamento dos dados provenientes das fontes selecionadas, mantendo-os o mais próximo possível de sua origem.

O levantamento realizado neste notebook ajudará a equipe a definir quais arquivos, tabelas e informações deverão ser considerados nessa etapa.

### 🥈 Camada Prata

Na camada Prata serão realizados os tratamentos necessários sobre os dados, como:

- seleção das variáveis necessárias;
- padronização;
- tratamento de valores ausentes;
- ajustes de tipos;
- filtros;
- preparação dos dados para os relacionamentos.

### 🥇 Camada Gold

A camada Gold será construída posteriormente com os dados tratados e organizados para responder às perguntas definidas para o projeto.

Nesta etapa poderão ser criados indicadores, relacionamentos e estruturas específicas para análise e visualização.

## Consulta de municípios - API IBGE

Para complementar as bases utilizadas no projeto, será utilizada a
API de Localidades do IBGE para obter a relação de municípios
brasileiros.

A API disponibiliza o código IBGE do município e seu respectivo nome,
permitindo relacionar essas informações ao campo `id_municipio`
presente nas tabelas da Base dos Dados.

Endpoint utilizado:

https://servicodados.ibge.gov.br/api/v1/localidades/municipios

A consulta não requer autenticação.

```python
import requests
import pandas as pd

url = "https://servicodados.ibge.gov.br/api/v1/localidades/municipios"

resposta = requests.get(url)
resposta.raise_for_status()

dados = resposta.json()

print(f"Quantidade de registros retornados: {len(dados)}")
```

```python
municipios = pd.DataFrame([
    {
        "id_municipio": str(municipio["id"]),
        "nome_municipio": municipio["nome"]
    }
    for municipio in dados
])

municipios.head()
```

```python
print("Quantidade de municípios:", len(municipios))
print("Códigos de municípios únicos:", municipios["id_municipio"].nunique())
```

```python
print("IDs nulos:", municipios["id_municipio"].isna().sum())
print("Nomes nulos:", municipios["nome_municipio"].isna().sum())
```

*Observação*

A API de Localidades do IBGE foi consultada para verificar a
disponibilidade de uma fonte oficial contendo o código e o nome
dos municípios brasileiros.

A consulta retornou 5.571 municípios, com 5.571 códigos únicos e
sem valores nulos para código ou nome do município.

Nesta etapa, os dados são utilizados exclusivamente para
levantamento e avaliação da fonte. A definição das colunas que
serão carregadas para a camada Bronze e os tratamentos que serão
realizados na camada Prata serão definidos posteriormente,
conforme a modelagem do projeto.
