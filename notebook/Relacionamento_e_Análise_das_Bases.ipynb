# 🔗 Relacionamento e Análise das Bases

## 1. Objetivo

Este notebook tem como objetivo definir as perguntas que orientarão as análises do projeto e identificar os dados necessários para respondê-las.

A partir do levantamento realizado no Notebook Analise Bases, serão avaliadas as relações entre a Base Principal de Avaliação da Alfabetização e as fontes complementares, buscando identificar:

- variáveis necessárias para cada análise;
- chaves e possibilidades de relacionamento entre as bases;
- informações que deverão ser preparadas na Camada Prata;
- indicadores e estruturas analíticas que poderão compor a Camada Gold.

As definições deste notebook servirão como base para a preparação dos dados e para a construção das análises e relatórios posteriores.

## Perguntas de análise

A análise será orientada por perguntas que permitam avaliar o alcance das metas de alfabetização, comparar o desempenho entre municípios e Unidades da Federação e investigar possíveis relações com características educacionais e socioeconômicas.

As perguntas foram organizadas considerando a Base Principal de Avaliação da Alfabetização e as fontes complementares levantadas no Notebook 1.

### 1. Alcance e evolução das metas

- Qual é a diferença entre o resultado de alfabetização observado e a meta estabelecida para cada município e período?
- Quais municípios atingiram ou superaram as metas estabelecidas?
- Como a distância entre resultado e meta evolui ao longo do período até 2030?

### 2. Comparação entre municípios e Unidades da Federação

- Quais Unidades da Federação apresentam os melhores resultados em relação às metas estabelecidas?
- Quais estados concentram maior quantidade ou proporção de municípios abaixo das metas?
- Existem diferenças relevantes no desempenho de municípios pertencentes a diferentes Unidades da Federação?

### 3. Contexto educacional

- Existe relação entre o desempenho de alfabetização e as características das escolas e dos alunos identificadas no Censo Escolar?
- Quais características educacionais apresentam maior associação com municípios que atingem ou superam as metas?

### 4. Contexto socioeconômico

- Existe relação entre o nível socioeconômico dos estudantes e o desempenho de alfabetização?
- Municípios com diferentes níveis de INSE apresentam diferenças relevantes no alcance das metas?

### 5. Análise integrada

- Quais municípios apresentam resultados de alfabetização acima ou abaixo do esperado quando considerados simultaneamente o cumprimento das metas, o contexto educacional e o contexto socioeconômico?
