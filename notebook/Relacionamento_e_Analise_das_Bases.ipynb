## 1. Objetivo

Este notebook tem como objetivo definir as perguntas que orientarão as análises do projeto e identificar os dados necessários para respondê-las.

A partir do levantamento realizado no Notebook Análise Bases, serão avaliadas as relações entre os dados da Avaliação da Alfabetização e as informações de identificação municipal e estadual, buscando identificar:

## Perguntas de análise

A análise será orientada por perguntas que permitam avaliar o desempenho da alfabetização, o alcance das metas e a evolução dos resultados ao longo do período analisado.

As perguntas foram definidas considerando os dados disponíveis na Base de Avaliação da Alfabetização e as informações de identificação e caracterização municipal obtidas a partir da API de Localidades do IBGE.

### 1. Indicador de alfabetização por município

- Qual é o indicador de alfabetização por município?

### 2. Alcance das metas

- Quais municípios atingiram ou superaram as metas estabelecidas?

### 3. Evolução do indicador

- Como o indicador de alfabetização evoluiu entre 2019 e 2025?

### 4. Distância entre resultado e meta

- Qual é a distância entre a meta estabelecida e o resultado observado por município?

### 5. Desigualdade regional

- Quais regiões concentram maior desigualdade nos resultados de alfabetização?

### 6. Desempenho por porte municipal

- Qual porte de município apresenta melhor desempenho de alfabetização?

## Relacionamentos entre as bases

Os relacionamentos serão definidos a partir das chaves de identificação disponíveis na Base de Avaliação da Alfabetização e na API de Localidades do IBGE.

O principal relacionamento utilizado será:

| Relação | Chave principal | Fonte relacionada | Finalidade |
|---|---|---|---|
| Município | `id_municipio` | Município / API IBGE | Identificar e relacionar os dados municipais |
| Unidade da Federação | `sigla_uf` | UF / API IBGE | Comparar os resultados em nível estadual |
| Região | identificação territorial do município | API IBGE | Agrupar e comparar resultados por região |
| Porte municipal | identificação territorial do município | API IBGE | Comparar o desempenho conforme o porte do município |

Os relacionamentos serão validados durante a preparação dos dados, considerando a compatibilidade dos identificadores e o nível de agregação das informações.
